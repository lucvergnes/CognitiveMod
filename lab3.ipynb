```python
def bayesFunction(param1, param2, param3):
    p_notH = 1.0 - param1
    p_D = param2 * param1 + param3 * p_notH
    output = (param1 * param2) / p_D
    return output
```

```python
print(bayesFunction(0.1, 0.9, 0.3))
print(bayesFunction(0.9, 0.9, 0.3))
print(bayesFunction(0.9, 0.3, 0.9))
```

```python
def bayesTheorem(priorH,likelihoodEH,likelihoodEnotH):
    notH = 1 - priorH
    HgivenE = priorH * likelihoodEH
    return ( HgivenE ) / ( HgivenE + ( notH * likelihoodEnotH) )
```

```python
print(bayesTheorem(0.1, 0.9, 0.3))
print(bayesTheorem(0.1, 0.9, 0.4))
print(bayesTheorem(0.9, 0.9, 0.3))
print(bayesTheorem(0.9, 0.3, 0.9))
```

```python
def bayesTheoremMulti(priorH,likelihoodEH):
    evidence = sum(
        likelihoodEH[h] * priorH[h] for h in priorH
    )

    # Posterior berekening
    posteriors = {
        h: (likelihoodEH[h] * priorH[h]) / evidence
        for h in priorH
    }

    return posteriors
```

```python
import numpy as np

def bayesFunctionMultipleHypotheses(priors, likelihoods):
    priors = np.array(priors)
    likelihoods = np.array(likelihoods)
    joint = priors * likelihoods
    evidence = np.sum(joint)
    posterior = joint[0] / evidence
    return posterior
```

```python

priorH2 = [0.3,0.3,0.4]
likelihood2 = [0.9,0.9,0.2]
print(bayesFunctionMultipleHypotheses(priorH2,likelihood2))
```

```python
def bayesFactor(priors1,posteriors1):
    priors = np.array(priors1)
    posteriors = np.array(posteriors1)

    priorOdds = priors[0] / sum(priors[1:])
    posteriorOdds = posteriors[0] / sum(posteriors[1:])

    bayesFac = [0]*len(posteriors)
    bayesFac[0] = priorOdds / posteriorOdds

    if len(bayesFac) >= 2:
        for i in range(1,len(bayesFac)):
            bayesFac[i] = (priors[0] / priors[i]) / (posteriors[0] / posteriors[i])

    return bayesFac
```

```python
def bayesFactor2(priors1,posteriors1):
    priors = np.array(priors1)
    posteriors = np.array(posteriors1)

    priorOdds = priors[0] / priors[1]
    posteriorOdds = posteriors[0] / posteriors[1]

    return priorOdds / posteriorOdds
```

```python
prior3 = [0.35,0.15,0.5]
likelihood3 = [0.3,0.4,0.3]
print(bayesFactor(prior3,likelihood3))
```
