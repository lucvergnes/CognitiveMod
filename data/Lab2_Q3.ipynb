```python
from sklearn.svm import SVC
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt
import numpy as np
```

```python
datafolder = r'C:/Users/Luc/PycharmProjects/CognitiveMod/data/'

neuralResonseS1 = pd.read_csv(datafolder + 'NeuralResponses_S1.txt')
neuralResonseS1
```

```python
results = pd.DataFrame(np.nan, index=range(88), columns=range(88))

for check1 in range(88):
    row1 = neuralResonseS1.iloc[check1]
    for check2 in range(88):

        row2 = neuralResonseS1.iloc[check2]
        corr = np.corrcoef(row1,row2)[0][1]
        invCorr = 1-corr
        results.loc[check1, check2] = invCorr

results
```

```python
plt.figure(figsize=(10, 8))
sb.heatmap(results, annot=False, cmap='viridis')
plt.xlabel('Response')
plt.ylabel('Response')
plt.title('1-correlation coefficient heatmap')
plt.show()
```
