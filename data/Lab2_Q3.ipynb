```python
from sklearn.svm import SVC
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt
import numpy as np
```

```python
datafolder = r'C:/Users/Luc/PycharmProjects/CognitiveMod/data/'

neuralResonseS1 = pd.read_csv(datafolder + 'NeuralResponses_S1.txt')
neuralResonseS1
```

```python
results = pd.DataFrame(np.nan, index=range(88), columns=range(88))

for check1 in range(88):
    row1 = neuralResonseS1.iloc[check1]
    for check2 in range(88):

        row2 = neuralResonseS1.iloc[check2]
        corr = np.corrcoef(row1,row2)[0][1]
        invCorr = 1-corr
        results.loc[check1, check2] = invCorr

results
```

```python
plt.figure(figsize=(10, 8))
sb.heatmap(results, annot=False, cmap='viridis')
plt.xlabel('Response')
plt.ylabel('Response')
plt.title('1-correlation coefficient heatmap')
plt.show()
```

```python
imageVectors = pd.read_csv(datafolder + 'CategoryVectors.txt')
imageVectorsAnimation = imageVectors['Var1']
print(len(imageVectorsAnimation))
imageVectorsAnimation
```

```python
animationMask = pd.DataFrame(np.nan, index=range(88), columns=range(88))

for check1 in range(88):
    animState1 = imageVectorsAnimation.iloc[check1]
    for check2 in range(88):
        animState2 = imageVectorsAnimation.iloc[check2]

        if animState1 == animState2:
            setRes = 0
        else:
            setRes = 1

        animationMask.loc[check1, check2] = setRes

animationMask
```

```python
plt.figure(figsize=(10, 8))
sb.heatmap(animationMask, annot=False, cmap='viridis')
plt.xlabel('Response')
plt.ylabel('Response')
plt.title('1-correlation coefficient heatmap')
plt.show()
```

```python
from scipy.stats import ttest_ind
from scipy.stats import pearsonr

sameDiss = []
diffDiss = []

for check1 in range(88):
    for check2 in range(check1,88):
        if check1 == check2:
            continue
        else:
            test1 = neuralResonseS1.iloc[check1]
            test2 = neuralResonseS1.iloc[check2]
            r, p = pearsonr(test1, test2)
            dissim = 1-r
            if animationMask.loc[check1, check2] == 0:
                sameDiss.append(dissim)
            else:
                diffDiss.append(dissim)
```

```python
t_stat, p_val = ttest_ind(sameDiss, diffDiss, alternative='two-sided')
print(str(t_stat) + '\t' + str(p_val))
print(p_val)
```

```python
print(len(sameDiss), len(diffDiss))
```

```python
from statistics import mean

avrgSame = mean(sameDiss)
avrgDiff = mean(diffDiss)
print(str(avrgSame) + '\t' + str(avrgDiff))
```

```python
std_dev1 = np.std(sameDiss, ddof=1)  # sample standard deviation
n1 = len(sameDiss)
std_error1 = std_dev1 / np.sqrt(n1)

std_dev2 = np.std(diffDiss, ddof=1)  # sample standard deviation
n2 = len(diffDiss)
std_error2 = std_dev2 / np.sqrt(n2)

print(str(std_error1) + '\t' + str(std_error2))
```

```python
barPlotData = pd.DataFrame({
    'Group': ['Similar', 'Different'],
    'Value': [avrgSame, avrgDiff],
    'SEM': [std_error1, std_error2]
})

fig, ax = plt.subplots(figsize=(10, 6))
sb.barplot(x='Group', y='Value', data=barPlotData, yerr=barPlotData['SEM'], capsize=0.5, errwidth=10)
plt.ylabel('Value')
plt.title('Barplot with SEM Error Bars')
plt.show()
```
