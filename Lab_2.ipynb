```python
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt
import numpy as np
```

```python
datafolder = r'C:/Users/Luc/PycharmProjects/CognitiveMod/data/'

neuralResonseS1 = pd.read_csv(datafolder + 'NeuralResponses_S1.txt')
neuralResonseS1
```

```python
neuralResAnimate = neuralResonseS1.iloc[:44].reset_index(drop=True)
neuralResInanimate = neuralResonseS1.iloc[44:88].reset_index(drop=True)
```

```python
AnimateAverageL = neuralResAnimate.mean(axis=1)
InanimateAverageL = neuralResInanimate.mean(axis=1)

boxPlotQ1 = pd.DataFrame({
    'Animate': AnimateAverageL,
    'Inanimate': InanimateAverageL
})

boxPlotQ1
```

```python
# Melt the DataFrame to long format
df_melted = pd.melt(boxPlotQ1)

plt.figure(figsize=(12, 8))
#sb.boxplot(x='variable', y='value', data=df_melted)
sb.barplot(x='variable', y='value', data=df_melted,errorbar="se")


plt.title("Neural response per category")
plt.xlabel("Category")
plt.ylabel("Values")

plt.show()
```

```python
AniGroup = boxPlotQ1['Animate']
InaniGroup = boxPlotQ1['Inanimate']
```

```python
from scipy.stats import ttest_ind

t_stat, p_value = ttest_ind(AniGroup, InaniGroup)
t_stat, p_value
```

```python
twentyAnimate = neuralResAnimate.iloc[:, :20]
twentyInanimate = neuralResInanimate.iloc[:, :20]

meanAnimate = twentyAnimate.mean()
meanInanimate = twentyInanimate.mean()
diffPlot = meanAnimate - meanInanimate
diffPlot
```

```python
x = np.arange(len(diffPlot))

plt.figure(figsize=(12, 6))  # Resize the plot (width=12, height=6 inches)
sb.barplot(x=x, y=diffPlot)

plt.xlabel('Voxel')
plt.ylabel('Response value')
plt.title('Neural response per voxel')
plt.show()
```
